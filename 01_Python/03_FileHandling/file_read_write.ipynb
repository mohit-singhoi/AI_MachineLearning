### File Handling in Python

File handling means: Handling a File 

    Creating, opening, reading, writing, and closing files to store data permanently.

👉 Data in variables is temporary

👉 Data in files is permanent

File handling in Python is used to read, write, append and manage files using the open() function and different modes.

**Types of File**

* 1.Text Files

         .txt, .csv

          Stores readable text

* 2.Binary Files

         .jpg, .pdf, .mp3

          Stores binary data

***1. file opening - open()***

***Syntax***

    file = open("filename", "mode")


##### File Models

| Mode | Meaning           |
| ---- | ----------------- |
| `r`  | Read              |
| `w`  | Write (overwrite) |
| `a`  | Append            |
| `x`  | Create new file   |
| `rb` | Read binary       |
| `wb` | Write binary      |
| `r+` | Read + write      |


```python
# If file does NOT exist → File is CREATED
# If file already exists → Old data is DELETED and file is opened
f = open("data.txt", "w")
f.write("Hello, World!\n")
f.write("This is a sample file.\n")
f.close()
```

***2. Reading a file***

read()

```python
# Example 
f = open("data.txt", "r")
content = f.read()
print(content)
f.close()
```

readline()

```python
# Example
f = open("data.txt", "r")
print(f.readline())
f.close()
```

readlines()

```python
# example
f = open("data.txt", "r")
lines = f.readlines()
print(lines)
f.close()
```

***3. Writing to File***

write()

```python
#Example
# w mode deletes old data
f = open("data.txt", "w")
f.write("Hello Python\n")
f.write("File Handling")
f.close()
```

append()

```python
# Example
# Old data remains
f = open("data.txt", "a")
f.write("\nNew line added")
f.close()
```

***4. Closing a File***

f.close()

why important?

* Frees memory

* Saves data properly

***5. with Statement***

* it  Automatically closes file
* No memory leak

```python
# Example
with open("data.txt", "r") as f:
    print(f.read())
```

***5. File Properties***

```python
# Example
f = open("data.txt", "r")
print(f.name)
print(f.mode)
f.close()
```

***6. Common error***

| Error             | Reason              |
| ----------------- | ------------------- |
| FileNotFoundError | File does not exist |
| PermissionError   | No permission       |
| IOError           | Input/Output issue  |


***7. Exception handling with file***

```python
# Example
try:
    f = open("abc.txt", "r")
    print(f.read())
except FileNotFoundError:
    print("File not found")
finally:
    print("Done")
    
```

```python
# File Handling Completed
# BIG FILE HANDLING PROGRAM

def write_file():
    with open("student.txt", "w") as f:
        f.write("Name: Mohit Kumar\n")
        f.write("Marks: 85\n")

def append_file():
    with open("student.txt", "a") as f:
        f.write("Grade: A\n")

def read_file():
    with open("student.txt", "r") as f:
        print(f.read())

def count_lines_words():
    with open("student.txt", "r") as f:
        lines = f.readlines()
        words = sum(len(line.split()) for line in lines)
        print("Lines:", len(lines))
        print("Words:", words)

write_file()
append_file()
read_file()
count_lines_words()
```

```python
# Example 7 :FILE HANDLING WITH EXCEPTION HANDLING - FULL PROGRAM

def file_operations():
    try:
        # Writing to file (file created if not exists)
        file = open("student1.txt", "w")
        file.write("Name: Mohit\n")
        file.write("Marks: 85\n")
        file.close()

        # Reading from file
        file = open("student.txt", "r")
        data = file.read()
        print("\n📄 File Content:")
        print(data)
        file.close()

    except FileNotFoundError:
        print("❌ Error: File not found")

    except PermissionError:
        print("❌ Error: Permission denied")

    except IOError:
        print("❌ Error: File input/output error")

    else:
        # Executes if no exception occurs
        print("✅ File operations completed successfully")

    finally:
        # Always executes
        print("🔚 Program finished")

# Function call
file_operations()
```
