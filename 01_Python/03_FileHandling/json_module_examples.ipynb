### JSON Module in Python

JSON (JavaScript Object Notation) is a lightweight data format used to store and exchange data.

* Human readable

* Language independent

* Very common in APIs, databases, web apps

#### JSON vs Python Data Types

| JSON         | Python       |
| ------------ | ------------ |
| object       | dict         |
| array        | list         |
| string       | str          |
| number       | int / float  |
| true / false | True / False |
| null         | None         |


**Why JSON Module**

Python provides json module to:

Convert Python → JSON

Convert JSON → Python

Read JSON file

Write JSON file

#### Important JSON Function

| Function       | Use                  |
| -------------- | -------------------- |
| `json.dumps()` | Python → JSON string |
| `json.loads()` | JSON string → Python |
| `json.dump()`  | Python → JSON file   |
| `json.load()`  | JSON file → Python   |


* json.dumps()  : (Python -> JSON STRING)

```python
# Example
import json

student = {
    "name": "Amit",
    "age": 21,
    "marks": 85
}

json_data = json.dumps(student)
print(json_data)
print(type(json_data))
```

* json.loads() : (JSON STRING -> PYTHON)

```python
#Example 
import json

data = '{"name":"Amit","age":21,"marks":85}'

py_data = json.loads(data)
print(py_data)
print(type(py_data))
```

* json.dump() : Writing json to file

```python
# Example
import json

student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python"
}

with open("student.json", "w") as f:
    json.dump(student, f)
```

json.load() : Reading JSON from File

```python
# Example
import json

with open("student.json", "r") as f:
    data = json.load(f)

print(data)
print(type(data))
```

***JSON with LIST of DICTIONARIES***

```python
# Example
import json

students = [
    {"id": 1, "name": "Amit"},
    {"id": 2, "name": "Neha"}
]

with open("students.json", "w") as f:
    json.dump(students, f)
```

***JSON Formatting***

json.dump(student, f, indent=4)

Makes JSON readable

***JSON with Exception Handling***

```python
# EXample 1
import json

try:
    with open("data.json", "r") as f:
        data = json.load(f)
        print(data)
except FileNotFoundError:
    print("File not found")
except json.JSONDecodeError:
    print("Invalid JSON format")
```

```python
# Example 2
# FULL JSON MODULE PROGRAM

import json

def write_json():
    student1 = {
        "name": "Mohit",
        "age": 23,
        "skills": ["Python", "ML", "Java"]
    }

    with open("student1.json", "w") as f:
        json.dump(student1, f, indent=4)

def read_json():
    try:
        with open("student1.json", "r") as f:
            data = json.load(f)
            print("📄 JSON Data:")
            print(data)
            print("Name:", data["name"])
            print("Skills:", data["skills"])

    except FileNotFoundError:
        print("❌ File not found")

    except json.JSONDecodeError:
        print("❌ Invalid JSON file")

    finally:
        print("🔚 JSON operation completed")

write_json()
read_json()
```
