### LAMBDA FUNCTION IN PYTHON

A lambda function is an anonymous function (function without a name) written in one single line.

* A lambda function is a small anonymous function defined using the lambda keyword.


#### Why Lambda Functions Are Used?

* Short & simple operations

* Used when a function is needed only once

* Mostly used with map(), filter(), reduce()


***Syntax***

lambda arguments : expression


***Rules***

No function name

Only one expression

Expression result is returned automatically

No return keyword

```python
# Normal Function

def square(x):
    return x * x
print(square(5))  # Output: 25
```

```python
# Lambda Function

square = lambda x: x * x
print(square(5))
# Output: 25
```

```python
# Lambda with Multiple arguments

add = lambda a, b: a + b
print(add(10, 20))
# Output: 30
```

```python
# Lambda with Conditional Expression

max_num = lambda a, b: a if a > b else b
print(max_num(10, 20))
# Output: 20
```

#### map() FUNCTION

map() applies a function to each element of a sequence.


***Syntax***

map(function, iterable)


```python
# Example 1 with Lambda

def square(x):
    return x*x

nums = [1, 2, 3, 4]
result = list(map(square, nums))
print(result)
```

```python
# Example 2 with Lambda

nums = [1, 2, 3, 4]
result = list(map(lambda x: x*x, nums))
print(result)
# Output: [1, 4, 9, 16]
```

#### filter() FUNCTION

filter() selects elements from a sequence based on a condition.


***Syntax***
filter(function, iterable)


```python
# Example 1 : Filter even numbers from a list using lambda
nums = [1, 2, 3, 4, 5, 6]
even_nums = list(filter(lambda x: x % 2 == 0, nums))
print(even_nums)
# Output: [2, 4, 6]
```

```python
# Example 2: Filter numbers greater than 3 from a list using lambda
nums = [1, 2, 3, 4, 5, 6]
greater_than_3 = list(filter(lambda x: x > 3, nums))
print(greater_than_3)
# Output: [4, 5, 6]
```

#### reduce FUNCTION

reduce() is in functools module

***Syntax***

from functools import reduce
reduce(function, iterable)



```python
# EXample : Sum of numbers in a list using lambda and reduce
from functools import reduce
nums = [1, 2, 3, 4, 5]
sum_nums = reduce(lambda x, y: x + y, nums)
print(sum_nums)
# Output: 15
```

#### One Line Cpmparison


| Function | Work                    |
| -------- | ----------------------- |
| map()    | Transforms each element |
| filter() | Selects elements        |
| reduce() | Reduces to single value |
