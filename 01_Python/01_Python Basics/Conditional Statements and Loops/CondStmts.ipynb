## Flow Control in Python 

Flow control in Python refers to the order in which statements are executed in a program.
By default, Python executes statements sequentially (top to bottom). Flow control statements allow the programmer to change this normal execution flow based on conditions or repetition of tasks.

**Flow control is mainly achieved using:**

1. Conditional Statements (Decision Making)

2. Loops (Iteration)



### 1. Conditional Statements 

Conditional statements are used to make decisions in a program. They execute different blocks of code depending on whether a given condition is True or False.

Python evaluates the condition and executes the corresponding block of code.

#### A. if Statement

The if statement executes a block of code only when a specified condition is True.

***Syntax :***

 if condition:

       statements


***Example:***

```python
age = 20
if age >= 18:
    print("Eligible for voting")
```

#### B. if-else Statement

The if–else statement executes one block of code when the condition is True and another block when the condition is False.


***Syntax:***

if condition:

      statements

else:

      statements


```python
# Example 

num = 7
if num % 2 == 0:
    print("Even")
else:
    print("Odd")

```

#### C.  if-elif-else Ladder :

The if–elif–else ladder is used to test multiple conditions sequentially. When a condition becomes True, its block is executed and the remaining conditions are skipped.


***Syntax***

if condition1:

      statements

elif condition2:

      statements
else:

      statements




```python
# Example

marks = 85
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
else:
    print("Grade C")
```

#### D. Nested if

A nested if statement means placing one if statement inside another if statement to check multiple related conditions.



```python
# Example

num = 10
if num > 0:
    if num % 2 == 0:
        print("Positive Even Number")
```
