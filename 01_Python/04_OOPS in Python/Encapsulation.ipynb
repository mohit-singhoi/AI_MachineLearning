### Encapsulation in Python

Encapsulation means wrapping data (variables) and methods into a single unit (class)
and restricting direct access to data.

Encapsulation is the process of binding data and methods together and restricting direct access.

* Data hiding

* Security

* Controlled access


**Why Encapsulation**

* Protects data from misuse

* Improves security

* Makes code maintainable

* Controls access using methods


**Access Modifiers in Python**

Python does not have strict access modifiers like Java, but it follows naming conventions:

| Type      | Syntax   | Meaning                        |
| --------- | -------- | ------------------------------ |
| Public    | `name`   | Accessible everywhere          |
| Protected | `_name`  | Accessible in class & subclass |
| Private   | `__name` | Accessible only inside class   |


**1. Public Members**

```python
# Example
class Student:
    def __init__(self, name):
        self.name = name   # public

s = Student("Mohit")
print(s.name)
```

**2. Proteted Members(_)**

```python
# Example
class Student:
    def __init__(self):
        self._college = "ABC College"

class Child(Student):
    def show(self):
        print(self._college)

c = Child()
c.show()

# d= Student() 
# print(d._college) // This is working but not Recommended

# Accessible in subclass
# Should not be accessed directly (convention)
```

**3. Private members(__)**

```python
# Example
class Student:
    def __init__(self):
        self.__marks = 90

    def show_marks(self):
        print(self.__marks)

s = Student()
s.show_marks()

# s.__marks -> Error
# Access only through methods
```

**Getter and Setter Methods**

Used to read and modify private data safely.

```python
# Example
class Bank:
    def __init__(self):
        self.__balance = 0

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            print("Invalid amount")

b = Bank()
b.set_balance(1000)
print(b.get_balance())
```

```python
# Full program
# BIG PROGRAM: ENCAPSULATION

class Employee:
    def __init__(self, name, salary):
        self.name = name          # public
        self._dept = "IT"         # protected
        self.__salary = salary   # private

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):
        if new_salary > self.__salary:
            self.__salary = new_salary
        else:
            print("Salary increment must be positive")

    def show_details(self):
        print("Name:", self.name)
        print("Department:", self._dept)
        print("Salary:", self.__salary)

emp = Employee("Mohit", 45000)

emp.show_details()
emp.set_salary(75000)
print("Updated Salary:", emp.get_salary())
```

***Imp Points***

* Encapsulation provides data hiding

* Private members use __

* Getters and setters control access

* Name mangling protects private data

* Improves security and maintainability

* Encapsulation → How data is protected

* Abstraction → What data is shown
