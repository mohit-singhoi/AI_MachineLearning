### OOPS In Python

OOPS (Object-Oriented Programming System) is a programming approach where we solve problems using objects.

👉 Real-world oriented

👉 Improves code reusability

👉 Easy maintenance


### CLasses & Objects in Python 


**Class :**

A class is a blueprint / template for creating objects.

👉 It defines:

     Variables (data)

     Methods (functions)

Real-life example:

     Class → Student

     Object → Rahul, Mohit, Neha

**Object :**

An object is a real instance of a class.

👉 Memory is allocated only when object is created.


***Syntax of class & Object***

class ClassName:

       # variables

       # methods

object_name = ClassName()


```python
# Example
class Student:
    def display(self):
        print("Welcome to OOPS")

s1 = Student()    # object creation
s1.display()

# self refers to current object

# Used to access class variables and methods

# Mandatory parameter in methods
```

**Class Variable vs Instance Variables**

* Instance Variable(Object- Level)

```python
# Example
class Student:
    def set_data(self, name, marks):
        self.name = name
        self.marks = marks

    def show(self):
        print(self.name, self.marks)

s1 = Student()
s1.set_data("Amit", 85)
s1.show()
```

* Class Variable ( Common for all Objects)

```python
# Example
class Student:
    college = "ABC College"   # class variable

    def show(self):
        print(Student.college)

s1 = Student()
s2 = Student()
s1.show()
s2.show()
print(Student.college)
```

* Constructor (___init___)

A constructor is a special method that:

* Executes automatically when object is created

* Used to initialize data

```python
# Example
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def show(self):
        print(self.name, self.marks)

s1 = Student("Rahul", 90)
s1.show()
s2 = Student("Amit", 85)
s2.show()
```

```python
# Complete Program 

# BIG PROGRAM: CLASS & OBJECTS

class Student:
    college = "Galgotias University"   # class variable

    def __init__(self, name, roll, marks):
        self.name = name          # instance variable
        self.roll = roll
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Roll:", self.roll)
        print("Marks:", self.marks)
        print("College:", Student.college)
        print("---------------------")

# Creating objects
s1 = Student("Amit", 1, 85)
s2 = Student("Mohit", 2, 90)

s1.display()
s2.display()
```
