### Abstraction In Python

Abstraction means hiding implementation details and
showing only essential features to the user.

* What to do ❌ How to do

* Focus on functionality

* Reduces complexity


**Phone Call Example:**

When a user makes a phone call, they only perform the action of dialing a number.
The user does not know how the call is connected internally, such as:

* Signal transmission

* Mobile towers

* Network switching

* Routing algorithms

All these internal processes are hidden from the user.

The user is concerned only with:

* Calling a number

* Talking to the other person

This is called abstraction, where what the system does is visible, but how it does it is hidden.


**Why Abstraction**

*Improves security*

*Reduces complexity*

*Enhances code readability*

*Supports loose coupling*


**How Abstraction is achieved in Python?**

Python achieves abstraction using:

1️⃣ Abstract Classes (abc module)

2️⃣ Abstract Methods

3️⃣ Interfaces (conceptually using ABC)


***1. Abstract Class :***

An abstract class is a class that:

* Cannot be instantiated

* May contain abstract methods


***2. Abstract Method***

An abstract method:

* Has no implementation

* Must be implemented in child class

```python
# Syntax
from abc import ABC, abstractmethod

class ClassName(ABC):
    @abstractmethod
    def method_name(self):
        pass
```

```python
# Example
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass  # Implement in future

class Rectangle(Shape):
    def area(self):
        print("Area of rectangle:", 10 * 5)

r = Rectangle()
r.area()
```
