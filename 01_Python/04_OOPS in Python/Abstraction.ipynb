### Abstraction In Python

Abstraction means hiding implementation details and
showing only essential features to the user.

* What to do ❌ How to do

* Focus on functionality

* Reduces complexity


**Phone Call Example:**

When a user makes a phone call, they only perform the action of dialing a number.
The user does not know how the call is connected internally, such as:

* Signal transmission

* Mobile towers

* Network switching

* Routing algorithms

All these internal processes are hidden from the user.

The user is concerned only with:

* Calling a number

* Talking to the other person

This is called abstraction, where what the system does is visible, but how it does it is hidden.


**Why Abstraction**

*Improves security*

*Reduces complexity*

*Enhances code readability*

*Supports loose coupling*


**How Abstraction is achieved in Python?**

Python achieves abstraction using:

1️⃣ Abstract Classes (abc module)

2️⃣ Abstract Methods

3️⃣ Interfaces (conceptually using ABC)


***1. Abstract Class :***

An abstract class is a class that:

* Cannot be instantiated

* May contain abstract methods


***2. Abstract Method***

An abstract method:

* Has no implementation

* Must be implemented in child class

```python
# Syntax
from abc import ABC, abstractmethod

class ClassName(ABC):
    @abstractmethod
    def method_name(self):
        pass
```

```python
# Example
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass  # Implement in future

class Rectangle(Shape):
    def area(self):
        print("Area of rectangle:", 10 * 5)

r = Rectangle()
r.area()
```

***Real-life Example***

ATM Machine:

You see → withdraw, deposit, balance check

You don’t see → bank server logic

That’s abstraction.

```python
# Example 2
from abc import ABC, abstractmethod

class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car started")

    def stop(self):
        print("Car stopped")

c = Car()
c.start()
c.stop()
```

```python
# BIG PROGRAM: ABSTRACTION

from abc import ABC, abstractmethod

class Bank(ABC):

    @abstractmethod
    def interest(self):
        pass

    def branch(self):
        print("Main Branch")

class UBGB(Bank):
    def interest(self):
        print("UBGB Interest Rate: 3%")

class SBI(Bank):
    def interest(self):
        print("SBI Interest Rate: 3.5%")

class HDFC(Bank):
    def interest(self):
        print("HDFC Interest Rate: 4.5%")

class Kotak(Bank):
    def interest(self):
        print("Kotak Mahindra Interest Rate: 7%")

banks = [UBGB(), SBI(), HDFC(),Kotak()]

for bank in banks:
    bank.interest()
    bank.branch()
```

#### Differentiate b/w Abstraction vs  Encapsulation

| Feature     | Abstraction      | Encapsulation    |
| ----------- | ---------------- | ---------------- |
| Focus       | What             | How              |
| Hides       | Implementation   | Data             |
| Achieved by | Abstract classes | Access modifiers |
| Level       | Design level     | Code level       |


***Imp Points***

* Abstract class cannot be instantiated

* Abstract methods have no body

* Child class must implement abstract methods

* Uses abc module

* Supports loose coupling

```python
# Baking System
from abc import ABC, abstractmethod

# =========================
# ABSTRACTION
# =========================
class Bank(ABC):

    def __init__(self, name):
        self.name = name   # public variable

    @abstractmethod
    def get_interest_rate(self):
        pass

    def bank_info(self):
        print("Welcome to", self.name)


# =========================
# INHERITANCE + POLYMORPHISM
# =========================
class SBI(Bank):

    def __init__(self, name, balance):
        super().__init__(name)
        self.__balance = balance   # ENCAPSULATION (private variable)

    # Method overriding (Polymorphism)
    def get_interest_rate(self):
        return "SBI Interest Rate: 5%"

    # Getter
    def get_balance(self):
        return self.__balance

    # Setter
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Amount deposited:", amount)
        else:
            print("Invalid amount")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Amount withdrawn:", amount)
        else:
            print("Insufficient balance")


# =========================
# OBJECT CREATION
# =========================
customer = SBI("State Bank of India", 10000)

customer.bank_info()                  # abstraction
print(customer.get_interest_rate())   # polymorphism
print("Balance:", customer.get_balance())

customer.deposit(2000)
customer.withdraw(3000)

print("Final Balance:", customer.get_balance())
```
