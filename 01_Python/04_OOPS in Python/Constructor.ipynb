### Constructor In Python

A constructor is a special method in Python that:

* Executes automatically when an object is created

* Used to initialize (assign) values to object variables

In Python, constructor name is __init__


***Syntax***

class ClassName:

     def __init__(self):

         # initialization code



***Object Creation***

obj = ClassName()

**Why Constructor is important**

Without constructor → values must be set manually

With constructor → values are set at object creation

***Example without Constructor***

```python
# Example
class Student:
    def set_data(self, name, marks):
        self.name = name
        self.marks = marks

    def show(self):
        print(self.name, self.marks)

s1 = Student()
s1.set_data("Mohit ", 93)
s1.show()
```

***Example WITH Constructor***

```python
# Example
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def show(self):
        print(self.name, self.marks)

s1 = Student("Amit", 85)
s1.show()
```

#### Types of Constructor

There are mainly two types of constructor

***1. Default Constructor***

* No Parameters (except self)

```python
# Example
class Demo:
    def __init__(self):
        print("Default constructor called")

d = Demo()
```

***2. Parameterized Constructor***

* Takes Parameters

```python
# Example
class Demo:
    def __init__(self, x, y):
        print(x + y)

d = Demo(10, 20)
d1 = Demo("Hello ", "World")
```

***Default Arguments***

```python
# Example
class Demo:
    def __init__(self, x=0, y=0):
        print(x + y)

Demo()
Demo(10)
Demo(10, 20)
```

***Constructor with Class Variable***

```python
# Example
class Student:
    college = "Galgotiya University"   # class variable

    def __init__(self, name):
        self.name = name

    def show(self):
        print(self.name, Student.college)

s1 = Student("Rahul")
s2 = Student("Mohit")
s1.show()
s2.show()
```

***Memory Concept***

* Constructor runs once per object

* Each object has separate memory

* Constructor initializes object state

```python
# Complete Program

class Employee:
    company = "Tech Solutions"   # class variable

    def __init__(self, emp_id, name, salary):
        self.emp_id = emp_id       # instance variables
        self.name = name
        self.salary = salary

    def display(self):
        print("ID:", self.emp_id)
        print("Name:", self.name)
        print("Salary:", self.salary)
        print("Company:", Employee.company)
        print("------------------------")

# Creating objects
e1 = Employee(101, "Amit", 50000)
e2 = Employee(102, "Neha", 60000)

e1.display()
e2.display()
```

***Imp. Points***

* Constructor is special method

* Name is __init__

* Executes automatically

* Initializes object variables

* Called once per object

* Python does not support constructor overloading

* A constructor in Python is a special method __init__ used to initialize object data automatically when an object is created.
