### Polymorphism in Python

Polymorphism means “one name, many forms”.

👉 Same function name

👉 Different behavior

Polymorphism allows the same method name to perform different tasks.

**Why Polymorphism is important**

* Improves code flexibility

* Supports scalability

* Reduces complexity

* Makes code reusable

#### Types of Polymorphism

Python supports 4 main types:

1️⃣ Function Polymorphism

2️⃣ Operator Overloading

3️⃣ Method Overriding

4️⃣ Duck Typing

**1. Function Polymorphism**

Same function works differently based on input.

```python
# Example
print(len("Python"))     # string
print(len([1, 2, 3]))    # list
print(len((1, 2)))       # tuple
```

**2. Operator Overloading**

Same operator behaves differently.

```python
# Example
print(10 + 20)        # addition
print("Py" + "thon")  # concatenation
print([1,2] + [3])    # list merge
```

**3. Method Overriding**

Child class provides its own implementation of parent method.

***Rules:***

* Same method name

* Same parameters

* Inherited class

```python
# Example
class Parent:
    def show(self):
        print("Parent method")

class Child(Parent):
    def show(self):
        print("Child method")

obj = Child()
obj.show()
```

```python
# Using super() with Overriding
class Parent:
    def show(self):
        print("Parent show")

class Child(Parent):
    def show(self):
        super().show()
        print("Child show")

c = Child()
c.show()
```

**4. Duck Typing**

Python doesn’t care about object type, only behavior.

Famous Line:

If it looks like a duck and quacks like a duck, it’s a duck.

```python
# Example
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

def make_sound(animal):
    animal.sound()

make_sound(Dog())
make_sound(Cat())
```

**Method Overloading**

Python does not support traditional method overloading

it achieved using default arguments

```python
# Example
class Calc:
    def add(self, a=0, b=0):
        print(a + b)

c = Calc()
c.add(10, 20)
c.add(5)
```

```python
# Full Program
# BIG PROGRAM: POLYMORPHISM

class Shape:
    def area(self):
        print("Area not defined")

class Rectangle(Shape):
    def area(self):
        print("Area of Rectangle:", 10 * 5)

class Circle(Shape):
    def area(self):
        print("Area of Circle:", 3.14 * 5 * 5)

def calculate_area(shape):
    shape.area()

r = Rectangle()
c = Circle()

calculate_area(r)
calculate_area(c)
```
