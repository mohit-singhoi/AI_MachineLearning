**Student Management System**

***Project Overview :***

The Student Management System is an OOPS-based Python application designed to manage student records efficiently. The system allows storing, updating, and displaying student information such as roll number, name, marks, and grade. It focuses on applying core Object-Oriented Programming concepts in a practical scenario.



***Objectives***

* To understand and implement OOPS concepts in Python

* To manage student data in a structured way

* To reduce manual record handling

* To ensure data security using encapsulation


***Technologies Used***

* Language: Python

* Paradigm: Object-Oriented Programming (OOPS)


***Key Features***

* Add new student records

* Update student marks

* Display student details with grades

* Automatic grade calculation

* Error handling for invalid input



***OOPS Concepts Applied***

* **Class & Object:** Used to represent students and their attributes

* **Constructor:** Initializes student data at the time of object creation

* **Encapsulation:** Student marks are kept private and accessed using methods

* **Inheritance:** Student class inherits from a base Person class

* **Polymorphism:** Method overriding used to display student details

* **Abstraction:** Abstract base class defines common structure for all users



***Outcome***

This project demonstrates a clear understanding of OOPS principles and shows how real-world problems can be solved using Python classes and objects. It is suitable for academic evaluation and entry-level software development roles.

```python
from abc import ABC, abstractmethod

# Abstraction
class Person(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def show_details(self):
        pass


# Inheritance
class Student(Person):
    def __init__(self, roll, name, marks):
        super().__init__(name)
        self.roll = roll
        self.__marks = marks   # encapsulation

    def update_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks")

    def grade(self):
        if self.__marks >= 75:
            return "A"
        elif self.__marks >= 60:
            return "B"
        else:
            return "C"

    # Polymorphism (method overriding)
    def show_details(self):
        print("Roll:", self.roll)
        print("Name:", self.name)
        print("Marks:", self.__marks)
        print("Grade:", self.grade())


# Main Program
try:
    students = []

    s1 = Student(1, "Mohit", 88)
    s2 = Student(2, "Rahul", 67)

    students.append(s1)
    students.append(s2)

    for s in students:
        s.show_details()
        print("------------")

    s1.update_marks(92)
    s1.show_details()

except Exception as e:
    print("Error:", e)
```
