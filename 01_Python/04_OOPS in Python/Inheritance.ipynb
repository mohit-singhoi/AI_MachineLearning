### Inheritance in Python

Inheritance is a mechanism where one class acquires the properties and methods of another class.

👉 Reusability of code

👉 Saves time

👉 Reduces redundancy

**Basic terminology**

| Term         | Meaning              |
| ------------ | -------------------- |
| Parent class | Base / Super class   |
| Child class  | Derived / Sub class  |
| Inheritance  | Acquiring properties |


***Syntax***

class Parent:

       pass

class Child(Parent):

       pass



```python
# Simple Example

class Animal:
    def sound(self):
        print("Animal makes sound")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

d = Dog()
d.sound()   # inherited method
d.bark()
```

**Types of Inheritance**

Python supports 5 types:

1️⃣ Single Inheritance

2️⃣ Multilevel Inheritance

3️⃣ Multiple Inheritance

4️⃣ Hierarchical Inheritance

5️⃣ Hybrid Inheritance

**1. Single Inheritance**

One parent -> one child

```python
# Example
class Parent:
    def show(self):
        print("Parent class")

class Child(Parent):
    def display(self):
        print("Child class")

c = Child()
c.show()
c.display()
```

**2. Multilevel inheritance**

parent -> Child -> Grandchild

```python
# Example
class A:
    def showA(self):
        print("Class A")

class B(A):
    def showB(self):
        print("Class B")

class C(B):
    def showC(self):
        print("Class C")

obj = C()
obj.showA()
obj.showB()
obj.showC()
```

**3. Multiple Inheritance**

One child -> multiple parents

```python
# Example
class Father:
    def skills(self):
        print("Construction Business")

class Mother:
    def skill(self):
        print("Cooking")

class Child(Father, Mother):
    pass

c = Child()
c.skills()   # Follows MRO
c.skill()
```

```python
# Method Resolution Order(MRO)
print(Child.mro())
```

**4. Hierarchical Inheritance**

One parent -> Multiple children

```python
# Example
class Parent:
    def show(self):
        print("Parent")

class Child1(Parent):
    pass

class Child2(Parent):
    pass

c1 = Child1()
c2 = Child2()
c1.show()
c2.show()
```

**5. Hybrid Inheritance**

Combination of multiple inheritance types

        A
       / \
      B   C
       \ /
        D


```python
# Hybrid Inheritance Example

class A:
    def showA(self):
        print("Class A")

class B(A):
    def showB(self):
        print("Class B")

class C(A):
    def showC(self):
        print("Class C")

class D(B, C):
    def showD(self):
        print("Class D")

# Object of class D
obj = D()

obj.showA()   # from class A
obj.showB()   # from class B
obj.showC()   # from class C
obj.showD()   # from class D
```

***Using super() keyword***

Used to call parent class constructor/method

```python
# Example
class Parent:
    def __init__(self):
        print("Parent constructor")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child constructor")

c = Child()
```

***Constructor in Inheritance***


```python
# Example
class Parent:
    def __init__(self):
        print("Parent")

class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child")

Child()
```

```python
# Inheritance of All types
# BIG PROGRAM: INHERITANCE IN PYTHON

class Person:
    def __init__(self, name):
        self.name = name

    def show_name(self):
        print("Name:", self.name)

class Student(Person):
    def __init__(self, name, roll):
        super().__init__(name)
        self.roll = roll

    def show_student(self):
        print("Roll:", self.roll)

class Result(Student):
    def __init__(self, name, roll, marks):
        super().__init__(name, roll)
        self.marks = marks

    def show_result(self):
        print("Marks:", self.marks)

# Object creation
r = Result("Mohit", 1, 90)

r.show_name()
r.show_student()
r.show_result()
```

***Imp. Points***

* Inheritance promotes reusability

* Parent class properties are inherited

* super() calls parent constructor

* Python supports multiple inheritance

* MRO decides method calling order
