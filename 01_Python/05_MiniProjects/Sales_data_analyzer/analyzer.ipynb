```python
import csv

class Sale:
    def __init__(self, sale_id, product, quantity, price):
        self.sale_id = sale_id
        self.product = product
        self.quantity = int(quantity)
        self.price = float(price)

    def total_price(self):
        return self.quantity * self.price

FILE_PATH = "Data/sales.csv"


class SalesAnalyzer:
    def __init__(self):
        self.sales = []
        self.load_sales()

    def load_sales(self):
        try:
            with open(FILE_PATH, "r") as file:
                reader = csv.DictReader(file)
                for row in reader:
                    sale = Sale(
                        row["sale_id"],
                        row["product"],
                        row["quantity"],
                        row["price"]
                    )
                    self.sales.append(sale)
        except FileNotFoundError:
            print("Sales file not found!")

    def total_sales(self):
        total = sum(sale.total_price() for sale in self.sales)
        print(f"Total Sales: ₹{total}")

    def average_sales(self):
        if len(self.sales) == 0:
            print("No sales data available")
            return
        avg = sum(sale.total_price() for sale in self.sales) / len(self.sales)
        print(f"Average Sale Value: ₹{avg:.2f}")

    def best_product(self):
        product_sales = {}

        for sale in self.sales:
            product_sales[sale.product] = product_sales.get(sale.product, 0) + sale.total_price()

        best = max(product_sales, key=product_sales.get)
        print(f"Best Selling Product: {best}")
        print(f"Total Revenue: ₹{product_sales[best]}")

    def sales_by_product(self):
        product_sales = {}

        for sale in self.sales:
            product_sales[sale.product] = product_sales.get(sale.product, 0) + sale.total_price()

        print("\nSales by Product:")
        for product, total in product_sales.items():
            print(f"{product} : ₹{total}")
```
