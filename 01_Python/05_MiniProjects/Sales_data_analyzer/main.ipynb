```python
# If analyzer.py doesn't exist, define SalesAnalyzer here
# Otherwise, uncomment the import below
# from analyzer import SalesAnalyzer

class SalesAnalyzer:
    def total_sales(self):
        print("Total Sales calculated")
    
    def average_sales(self):
        print("Average Sale Value calculated")
    
    def best_product(self):
        print("Best Selling Product identified")
    
    def sales_by_product(self):
        print("Sales By Product displayed")


def main():
    analyzer = SalesAnalyzer()

    while True:
        print("\n===== Sales Data Analyzer =====")
        print("1. Total Sales")
        print("2. Average Sale Value")
        print("3. Best Selling Product")
        print("4. Sales By Product")
        print("5. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            analyzer.total_sales()
        elif choice == "2":
            analyzer.average_sales()
        elif choice == "3":
            analyzer.best_product()
        elif choice == "4":
            analyzer.sales_by_product()
        elif choice == "5":
            print("Exiting...")
            break
        else:
            print("Invalid choice!")


if __name__ == "__main__":
    main()
```
