### List in Python

A list is a built-in, ordered, and mutable data structure in Python used to store multiple elements in a single variable.


**Characteristics of List**

* Ordered – elements have fixed positions

* Mutable – elements can be changed

* Indexed – index starts from 0

* it Allows duplicates

* Heterogeneous – different data types allowed

* Dynamic size – can grow or shrink


***Creating a List***

```python
# Eample of Creating lists in Python

a = [10, 20, 30]
b = ["apple", "banana"]
c = [1, "hi", 3.5, True]
d = list((1, 2, 3))   # using list()
```

```python
# Printing the lists
print(a)
print(b)
print(c)
print(d)
```

***Indexing in List***

```python
# Indexing in lists
nums = [10, 20, 30, 40]

nums[0]    # 10
nums[-1]   # 40
nums[2]    # 30
nums[0] = 15   # Modifying the first element
print(nums)    # [15, 20, 30, 40]

nums[0]

# Negative index starts from -1 (last element)
```

***Slicing in List***


```python
# Example of Slicing in lists

numbers = [10, 20, 30, 40, 50, 60, 70]
slice1 = numbers[1:5]    # [20, 30, 40, 50]
slice2 = numbers[:4]      # [10, 20, 30, 40]
slice3 = numbers[3:]      # [40, 50, 60, 70]
slice4 = numbers[-4:-1]   # [40, 50, 60]
slice5 = numbers[:]       # [10, 20, 30,
```

```python
# printing slices
print(slice1)
print(slice2)
print(slice3)
print(slice4)
print(slice5)
```

***Mutability***

List is Mutable in Nature

```python
nums = [10, 20, 30]
nums[1] = 25
print(nums)
# [10, 25, 30]
```

#### All List Methods

***1. append()***

Adds one element at the end

```python
# List Append
fruits = ["apple", "banana"]
fruits.append("cherry")
print(fruits)

# ['apple', 'banana', 'cherry']


a = [1, 2, 3]
a.append(4)
print(a)
# [1, 2, 3, 4]
```

***2. extend()***

Adds multiple elements

```python
# Example list extend
list1 = [1, 2, 3]
list2 = [4, 5, 6]
list1.extend(list2)
print(list1)
# [1, 2, 3, 4, 5, 6]
```

***3. insert(index , value)***

Inserts element at specific position

```python
# Example insert(index , value)
colors = ["red", "blue", "green"]
colors.insert(1, "yellow")
print(colors)
# ['red', 'yellow', 'blue', 'green']
```

***4. remove(value)***

Removes first occurrence of value

```python
# Remove item by value

colors.remove("blue")
print(colors)
# ['red', 'yellow', 'green']
```

***5. pop()***

Removes and returns element

```python
# Example  pop() method

a.pop()      # removes last
a.pop(1)     # removes index 1
```

```python
# Example  pop() method
numbers = [10, 20, 30, 40, 50]
last_item = numbers.pop()    # removes and returns 50
print(last_item)             # 50
print(numbers)               # [10, 20, 30, 40]
```

***6. clear()***

Remove all elements

```python
# Remove item using clear() method

a.clear()    # removes all items
print(a)      # []
```

***7. index(value)***

Returns index of element

```python
# example of index() method
items = ['a', 'b', 'c', 'd']
index_c = items.index('c')   # 2
print(index_c)
# 2
```

```python
# example
nums = [10, 20, 30]
nums.index(20)
```

***8. Count()***

Count Occurrences

```python
# Example count() method
values = [1, 2, 2, 3, 2, 4]
count_2 = values.count(2)
print(count_2)
# 3
```

***9. reverse()***

Reverse list order

```python
# Example of reverse() method
data = [1, 2, 3, 4, 5]
data.reverse()
print(data)
# [5, 4, 3, 2, 1]
```

***10. sort()***

Sort list (ascending by default)

```python
# Example of sort() method
marks = [88, 45, 67, 92, 75]
marks.sort()
print(marks)
# [45, 67, 75, 88, 92]
```

***11. copy()***

Creates shallow copy

```python
# Example of list copy() method
original = [1, 2, 3]
copy_list = original.copy()
print(copy_list)
# [1, 2, 3]
```

***12. max() , min() , sum()***

It return max , min, sum of the list

```python
# Example max() , min() , sum()
nums = [10, 25, 5, 30, 15]
print(max(nums))  # Output: 30
print(min(nums))  # Output: 5
print(sum(nums))  # Output: 85
```

***13. Len()***

It return the length of the sets

```python
# Length of list
my_list = [1, 2, 3, 4, 5]
length = len(my_list)
print(length)  # Output: 5
```

### Looping in List

1. For Loop

2. While Loop

***1. for loop***

```python
# For Loop through a list
colors = ["red", "green", "blue"]
for color in colors:
    print(color)
    
```

```python
# Loop through list with index
nums = [10, 20, 30, 40]
for x in nums:
    print(x)
```

***2. while loop***

```python
# List Example of while loop
numbers = [1, 2, 3, 4, 5]
i = 0
while i < len(numbers):
    print(numbers[i])
    i += 1
    
```
