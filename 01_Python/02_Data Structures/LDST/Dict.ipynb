### Dictionary in Python

A dictionary is a collection of key–value pairs.

***Key Characteristics***

* Keys are unique

* Values can be duplicate

* Mutable data type

* Unordered (in older versions) / Insertion-ordered (Python 3.7+)

* Keys must be immutable (int, float, string, tuple)

* Values can be of any data type

***Syntax***

d = {key: value}


***1. Creating a Dictionary***

```python
# Example Creating a Dictionary
student = {
    "name": "Mohit",
    "age": 21,
    "course": "MCA"
}

print(student)


d = {}
print(d)  # Output: {}
```

***2. Accessing Dictinary Elements***

```python
# Using key
print(student["name"])   # Mohit


# Using get() method
print(student.get("age"))  # 21
```

***3. Adding and Updating Elements***

```python
student["marks"] = 85        # Add
student["age"] = 22          # Update

print(student)
```

***4. Removing Elements***

| Method      | Description                |
| ----------- | -------------------------- |
| `pop(key)`  | Removes specified key      |
| `popitem()` | Removes last inserted item |
| `del`       | Deletes key                |
| `clear()`   | Removes all items          |


```python
# Removing element
student.pop("marks")
student.popitem()
del student["age"]
student.clear()
```

***5. Looping Throught Dictionary***

```python
# Example 
student = {"name": "Rahul", "age": 21, "course": "MCA"}

for key in student:
    print(key, student[key])

for key, value in student.items():
    print(key, value)
```

***6. Dictionary Built-in Methods***

| Method         | Description                        |
| -------------- | ---------------------------------- |
| `keys()`       | Returns all keys                   |
| `values()`     | Returns all values                 |
| `items()`      | Returns key-value pairs            |
| `get()`        | Returns value for key              |
| `update()`     | Updates dictionary                 |
| `pop()`        | Removes key                        |
| `popitem()`    | Removes last item                  |
| `clear()`      | Removes all                        |
| `copy()`       | Shallow copy                       |
| `setdefault()` | Returns value, adds key if missing |
| `fromkeys()`   | Create dictionary from keys        |


***7. Example of Dictionary Methods***

key() , values() , items()

```python
# Example  of key() , values() , items()
d = {"a": 1, "b": 2}

print(d.keys())
print(d.values())
print(d.items())
```

```python
# Update 
d1 = {"a": 1}
d2 = {"b": 2}
d1.update(d2)
print(d1)
```

```python
# setdefault
d = {"a": 1}
d.setdefault("b", 2)
print(d)
```

```python
# fromkeys
keys = ["id", "name", "age"]
d = dict.fromkeys(keys, 0)
print(d)
```

***8. Dictionary Membership Test***


```python
# Example
d = {"name": "Amit", "age": 22}

print("name" in d)      # True
print("marks" not in d) # True
```

***9. Nested Dictionary***

```python
# Example
student = {
    "name": "Amit",
    "marks": {"math": 80, "cs": 90}
}

print(student["marks"]["cs"])  # 90
```

***10. Dictionary Comprehension***

```python
# Example  --- IGNORE ---
squares = {x: x*x for x in range(1, 6)}
print(squares)
```

#### Dictionary vs List vs Tuple

| Feature     | Dictionary     | List     | Tuple    |
| ----------- | -------------- | -------- | -------- |
| Storage     | Key–Value      | Elements | Elements |
| Mutable     | Yes            | Yes      | No       |
| Indexing    | By key         | By index | By index |
| Speed       | Fast lookup    | Medium   | Fast     |
| Key allowed | Immutable only | N/A      | N/A      |


**Summary :**

* Dictionary stores key–value pairs

* Keys must be unique and immutable

* Dictionary is mutable

* get() is safer than []

* Dictionary is fast for searching

* Supports nesting and comprehension
