### Sets in Python

A set in Python is a collection of unique items.

* Unordered: Elements have no index.

* Mutable: You can change the set (add/remove elements), except frozenset.

* Unique: No duplicate elements allowed.

* Immutable elements only: Cannot store lists, dictionaries, or other sets, but can store numbers, strings, and tuples.


##### Uses of Sets:

* Remove duplicates from a list

* Perform mathematical operations like union, intersection

* Fast membership testing (in / not in)

***1. Creating Sets*** 

```python
# Example of creating sets in Python

# Using curly braces
my_set = {1, 2, 3, 4}

# Using set() function
my_set = set([1, 2, 3, 3, 4])
print(my_set)  # {1, 2, 3, 4} duplicates removed

# Empty set
empty_set = set()
print(empty_set)  # set()
```

***2. Accessing Set Elements***

Since sets are unordered, you cannot access elements by index.

```python
# Example of Accessing Set Elements in Python
my_set = {10, 20, 30, 40, 50}
for element in my_set:
    print(element)
# Note: Sets are unordered, so elements may appear in any order
```

***3. Adding Elements***

| Method             | Description            | Example           |
| ------------------ | ---------------------- | ----------------- |
| `add(elem)`        | Adds a single element  | `s.add(5)`        |
| `update(iterable)` | Adds multiple elements | `s.update([6,7])` |


```python
# Example of Adding Elements to a Set in Python
my_set = {1, 2, 3}
my_set.add(4)
print(my_set)  # {1, 2, 3, 4}

my_set.update([5, 6])
print(my_set)  # {1, 2, 3, 4, 5, 6}
my_set.update({7, 8})
print(my_set)  # {1, 2, 3, 4, 5, 6, 7, 8}
```

***4. Removing Elements***

| Method          | Description                           | Example         |
| --------------- | ------------------------------------- | --------------- |
| `remove(elem)`  | Remove element, error if not found    | `s.remove(2)`   |
| `discard(elem)` | Remove element, no error if not found | `s.discard(10)` |
| `pop()`         | Removes **random element**            | `x = s.pop()`   |
| `clear()`       | Remove all elements                   | `s.clear()`     |


```python
# Example of removing Elements from a Set in Python
my_set = {1, 2, 3, 4, 5}
my_set.remove(3)  # Raises KeyError if 3 not found
print(my_set)  # {1, 2, 4, 5}

my_set.discard(4)  # Does not raise error if 4 not found
print(my_set)  # {1, 2, 5}

popped_element = my_set.pop()  # Removes and returns an arbitrary element
print(popped_element)
print(my_set)

my_set.clear()  # Empties the set
print(my_set)  # set()
```

***5. Set Operations( Mathematical)***

| Operation            | Symbol | Method                   | Example                     |        |                  |
| -------------------- | ------ | ------------------------ | --------------------------- | ------ | ---------------- |
| Union                | `      | `                        | `union()`                   | `{1,2} | {2,3} → {1,2,3}` |
| Intersection         | `&`    | `intersection()`         | `{1,2} & {2,3} → {2}`       |        |                  |
| Difference           | `-`    | `difference()`           | `{1,2,3} - {2,3} → {1}`     |        |                  |
| Symmetric Difference | `^`    | `symmetric_difference()` | `{1,2,3} ^ {2,3,4} → {1,4}` |        |                  |


```python
# Example of Set Operations in Python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# Union
union_set = set_a.union(set_b)
print(union_set)  # {1, 2, 3, 4, 5, 6}

# Intersection
intersection_set = set_a.intersection(set_b)
print(intersection_set)  # {3, 4}

# Difference
difference_set = set_a.difference(set_b)
print(difference_set)  # {1, 2}

# Symmetric Difference
symmetric_difference_set = set_a.symmetric_difference(set_b)
print(symmetric_difference_set)  # {1, 2, 5, 6}
```

```python

# Example of Set Methods in Python
my_set = {1, 2, 3, 4, 5}
print(len(my_set))  # 5
print(3 in my_set)  # True
print(6 not in my_set)  # True
my_set_copy = my_set.copy()
print(my_set_copy)  # {1, 2, 3, 4, 5}
```

```python
# Example of Set Comprehension in Python
squared_set = {x**2 for x in range(1, 6)}
print(squared_set)  # {1, 4, 9, 16, 25}
```

***6. Membership Testing***

```python
# Membership Testing in Sets
fruits = {"apple", "banana", "cherry"}
print("apple" in fruits)    # True
print("mango" not in fruits)  # True
print("banana" in fruits)   # True
print("grape" not in fruits)  # True
```

***7. Set Relationships***

| Method              | Description                              | Example                            |
| ------------------- | ---------------------------------------- | ---------------------------------- |
| `issubset(other)`   | Returns True if set is subset of other   | `{1,2}.issubset({1,2,3}) → True`   |
| `issuperset(other)` | Returns True if set is superset of other | `{1,2,3}.issuperset({1,2}) → True` |
| `isdisjoint(other)` | Returns True if no common elements       | `{1,2}.isdisjoint({3,4}) → True`   |


```python
# Example of Set Realationship Methods in Python

set_x = {1, 2, 3}
set_y = {1, 2, 3, 4, 5}

print(set_x.issubset(set_y))  # True
print(set_y.issuperset(set_x))  # True
print(set_x.isdisjoint({6, 7, 8}))  # True
```

***8. Copying a Set***

copy() → Shallow copy of set


```python
# EXample of Copying a Set in Python
original_set = {10, 20, 30}
copied_set = original_set.copy()
print(copied_set)  # {10, 20, 30}

# 2nd example
s1 = {1,2,3}
s2 = s1.copy()
print(s2)  # {1,2,3}
```

***9. Frozenset(immutable Set)***

* Frozenset is immutable → cannot be modified

* Can be used as dictionary keys

```python
# frozenset example
frozen_set = frozenset([1, 2, 3, 4])
print(frozen_set)  # frozenset({1, 2, 3, 4})


fs = frozenset([1, 2, 3])
print(fs)
# fs.add(4) -> Error
# fs.remove(2) -> Error
```
