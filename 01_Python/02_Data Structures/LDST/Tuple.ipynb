### Tuples in Python

A tuple is a collection of ordered and immutable elements.

***Key Characteristics***

✅ Ordered – elements have a fixed position (index)

❌ Immutable – elements cannot be changed after creation

✅ Allows duplicates

✅ Can store different data types

✅ Faster than lists (because immutable)

✅ Can be used as dictionary keys (if elements are immutable)

***1. Creating a Tuple***



```python
# Creation of Tuple in different way

t1 = (10, 20, 30)
t2 = 10, 20, 30        # without parentheses
t3 = ("apple", "banana", "cherry")
t4 = (1, "Python", 3.5)
t5 = tuple([1, 2, 3, 4])  # using tuple() constructor

# Single Element tuple
t = (5,)     # comma is mandatory
print(t)  # Output: (5,)
```

***2. Accessing Tuple Elements***

```python
# Using indexing to access elements
my_tuple = (10, 20, 30, 40, 50)
print(my_tuple[0])    # Output: 10
print(my_tuple[2])    # Output: 30
print(my_tuple[-1])   # Output: 50


# using slicing to access elements
my_tuple = (10, 20, 30, 40, 50)
print(my_tuple[1:4])   # Output: (20, 30,
# Output: (20, 30, 40)
print(my_tuple[:3])    # Output: (10, 20, 30)
# Output: (10, 20, 30)
print(my_tuple[2:])    # Output: (30, 40, 50)
# Output: (30, 40, 50)
```

***3. Tuple is Immutable***

You cannot modify a tuple


```python
# Example 1
t = (1, 2, 3)
# t[0] = 10   ❌ TypeError


# Example 2
t = (10, 20, 30)
# del t[1]    ❌ TypeError
```

```python
# Example : Tuple immutable 
# But you can convert it to a list:
t = (1, 2, 3)
l = list(t)
l[0] = 10
t = tuple(l)
print(t)  # Output: (10, 2, 3)
```

***4. Looping Through Tuple***


```python
# Example:1
t = (10, 20, 30)
for i in t:
    print(i)
    
# Example:2    
    
for i in range(len(t)):
    print(t[i])
```

***5. Tuple Operation***

```python
# Concatatenation of Tuples
t1 = (1, 2, 3)
t2 = (4, 5, 6)
t3 = t1 + t2
print(t3)  # Output: (1, 2, 3, 4, 5, 6)


# Repetition of Tuples
t = (1, 2, 3)
t_repeated = t * 3
print(t_repeated)  # Output: (1, 2, 3, 1, 2, 3, 1, 2, 3)


# Tuple unpacking
t = (10, 20, 30)
a, b, c = t
print(a)  # Output: 10
print(b)  # Output: 20
print(c)  # Output: 30

# Membership Testing in Tuples
fruits = ("apple", "banana", "cherry")
print("apple" in fruits)    # True
print("mango" not in fruits)  # True
print("banana" in fruits)   # True
print("grape" not in fruits)  # True
```

***6. Built-in Function with Tuple***

```python
# Example of Tuple Built-in Functions
t = (10, 20, 30, 40)

print(len(t))   # 4
print(max(t))   # 40
print(min(t))   # 10
print(sum(t))   # 100


# Count() and Index() Methods
t = (1, 2, 3, 2, 4, 2)
print(t.count(2))  # 3
print(t.index(3))  # 2
```

***7. Tuple Packing and Unpacking***

```python
# Example of packing and unpacking
packed_tuple = 1, "Hello", 3.14  # Packing
a, b, c = packed_tuple            # Unpacking
print(a)  # Output: 1
print(b)  # Output: Hello
print(c)  # Output: 3.14


# using * operator
t = (1, 2, 3, 4, 5)
a, *b, c = t
print(a)  # 1
print(b)  # [2, 3, 4]
print(c)  # 5
```

***10. Nexted Tuple***

```python
# Nested Tuples
nested_tuple = (1, (2, 3), (4, (5, 6)))
print(nested_tuple[1])        # Output: (2, 3)
print(nested_tuple[2][1])     # Output: (5, 6)
print(nested_tuple[2][1][0])  # Output: 5
```

#### Tuple Vs List

| Feature          | List          | Tuple       |
| ---------------- | ------------- | ----------- |
| Nature           | Mutable       | Immutable   |
| Syntax           | `[]`          | `()`        |
| Modification     | Allowed       | Not allowed |
| Duplicate values | Allowed       | Allowed     |
| Ordering         | Ordered       | Ordered     |
| Speed            | Slower        | Faster      |
| Memory usage     | More          | Less        |
| Built-in methods | Many          | Very few    |
| Safety           | Less secure   | More secure |
| Dictionary key   | ❌ Not allowed | ✅ Allowed   |
| Use case         | Dynamic data  | Fixed data  |


#### Why use Tuple

* Protect data from modification

* Faster execution

* Used for fixed collections

* Used as dictionary keys

* Used in function returns
