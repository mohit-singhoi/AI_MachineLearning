### List Comprehension in Python

List comprehension is a short and powerful way to create lists in Python using a single line of code instead of using loops.

***Why use list comprehension?***

Shorter code

More readable

Faster execution

Cleaner logic

***1. Normal List Creation Vs List Comprehension***

```python
# Using for loop
squares = []
for i in range(1, 6):
    squares.append(i * i)

print(squares)
```

```python
# Using list comprehension
squares_comp = [i * i for i in range(1, 6)]
print(squares_comp)
```
