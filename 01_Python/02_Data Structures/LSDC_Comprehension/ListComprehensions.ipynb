### List Comprehension in Python

List comprehension is a short and powerful way to create lists in Python using a single line of code instead of using loops.

***Why use list comprehension?***

Shorter code

More readable

Faster execution

Cleaner logic

**Basic Syntax**

new_list = [expression for item in iterable]


```python
#Example
numbers = [1, 2, 3, 4]
result = [x * 2 for x in numbers]
print(result)
```

**List Comprehension with condition**

**Syntax**

new_list = [expression for item in iterable if condition]


```python
evens = [x for x in range(1, 11) if x % 2 == 0]
print(evens)
```

***1. Normal List Creation Vs List Comprehension***

```python
# Using for loop
squares = []
for i in range(1, 6):
    squares.append(i * i)

print(squares)
```

```python
# Using list comprehension
squares_comp = [i * i for i in range(1, 6)]
print(squares_comp)
```

***2. List Creation with if-else***

***Syntax***

new_list = [expression_if_true if condition else expression_if_false for item in iterable]


```python
# Example
result = ["Even" if x % 2 == 0 else "Odd" for x in range(1, 6)]
print(result)
```

***3. Nested List Comprehension***

***Syntax***

[expression for item1 in iterable1 for item2 in iterable2]


```python
# Example
pairs = [(x, y) for x in [1, 2] for y in [3, 4]]
print(pairs)
```

***4. List Comprehension with String***

```python
# Example
word = "python"
letters = [ch.upper() for ch in word]
print(letters)

# Example
filtered_squares = [x * x for x in range(1, 11) if x % 2 != 0]
print(filtered_squares)
```

***5. List Comprehension with Function***

```python
# Example
def square(n):
    return n * n

result = [square(x) for x in range(1, 6)]
print(result)
```

***6. List Comprehension with List of List***

```python
# Example
matrix = [[1, 2], [3, 4], [5, 6]]

flatten = [item for sublist in matrix for item in sublist]
print(flatten)
```

***7. List Comprehension with Dictionary***

```python
# Example
data = {"a": 1, "b": 2, "c": 3}

keys = [k for k in data]
values = [v for v in data.values()]

print(keys)
print(values)
```

***8. List Comprehension vs Loop***

| Feature     | Loop   | List Comprehension |
| ----------- | ------ | ------------------ |
| Code length | Long   | Short              |
| Readability | Medium | High               |
| Speed       | Slower | Faster             |
| Usage       | Basic  | Advanced           |


***9. Practice Qns***

```python
# Qns -1 Squares of even numbers from 1 to 20
even_squares = [x * x for x in range(1, 21) if x % 2 == 0]
print(even_squares)

# Qns -2 First letter of each word in a sentence
sentence = "List comprehensions are powerful"
first_letters = [word[0] for word in sentence.split()]
print(first_letters)

# Qns -3 Flatten a list of lists
list_of_lists = [[1, 2], [3, 4], [5, 6]]
flattened = [item for sublist in list_of_lists for item in sublist]
print(flattened)


# Qns -4 Lengths of words longer than 3 letters
words = ["I", "love", "programming", "in", "Python"]
lengths = [len(word) for word in words if len(word) > 3]
print(lengths)


# Qns -5 Uppercase vowels in a string
text = "Hello World"
vowels_upper = [ch.upper() for ch in text if ch.lower() in 'aeiou']
print(vowels_upper)


# Qns -6 Tuples of numbers and their cubes
cubes = [(x, x ** 3) for x in range(1, 6)]
print(cubes)


# Qns -7 Filter positive numbers and square them
numbers = [-10, 15, -20, 25, 30]
positive_squares = [x * x for x in numbers if x > 0]
print(positive_squares)


# Qns -8 Create a list of booleans indicating even numbers
nums = [1, 2, 3, 4, 5]
is_even = [x % 2 == 0 for x in nums]
print(is_even)


# Qns -9 Convert temperatures from Celsius to Fahrenheit
celsius = [0, 20, 37, 100]
fahrenheit = [(temp * 9/5) + 32 for temp in celsius]
print(fahrenheit)


# Qns -10 Create a list of strings with "Number: x" format
numbers = [1, 2, 3, 4, 5]
formatted = [f"Number: {x}" for x in numbers]
print(formatted)


# Qns -11 Square of num
nums = [1, 2, 3, 4, 5]
squares = [x*x for x in nums]
print(squares)


# Qns - 12 Filter Positive Numbers
nums = [-5, 3, -1, 6, 0]
positive = [x for x in nums if x > 0]
print(positive)


# Qns - 13  Convert List to Upper Case
names = ["ram", "shyam", "mohan"]
upper_names = [name.upper() for name in names]
print(upper_names)

# Qns - 14  Create Tuples of Numbers and Their Squares
nums = [1, 2, 3, 4, 5]
num_squares = [(x, x*x) for x in nums]
print(num_squares)

# Qns - 15 Convert List to upper case if length > 3
words = ["hi", "hello", "hey", "greetings"]
upper_words = [word.upper() for word in words if len(word) > 3]
print(upper_words)
```

***Imp. Points***

* List comprehension is a compact form of loop

* Uses expression + for + optional if

* Faster than traditional loops

* Supports conditions and nesting

* Improves readability

* Used for filtering and transformation
