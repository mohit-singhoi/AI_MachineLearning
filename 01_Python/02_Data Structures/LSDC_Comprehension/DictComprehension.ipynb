### Dictionary Comprehension in Python

Dictionary comprehension is a short and elegant way to create dictionaries using a single line of code instead of loops.

* Generate key–value pairs

* Apply conditions

* Transform data easily

Just like list comprehension, but for dictionaries.

#### Normal Dictionary Creation vs Dictionary Comprehension

```python
# Using for loop
squares = {}
for i in range(1, 6):
    squares[i] = i * i

print(squares)
```

```python
# Using dictionary comprehension
squares = {i: i*i for i in range(1, 6)}
print(squares)
```

***Basic Syntax***

new_dict = {key_expression : value_expression for item in iterable}


```python
# Example
d = {x: x*x for x in range(1, 5)}
print(d)
```

***1. Dictionary Comprehension With Condition***

***Syntax***
  {key: value for item in iterable if condition}


```python
# Example Even Number as a key
evens = {x: x*x for x in range(1, 11) if x % 2 == 0}
print(evens)
```

***2. Dictionary Comprehension with if-else***

***Syntax***

   {key: value_if_true if condition else value_if_false for item in iterable}



```python
# Exampleresult = {x: "Even" if x % 2 == 0 else "Odd" for x in range(1, 6)}
result = {x: "Even" if x % 2 == 0 else "Odd" for x in range(1, 6)}
print(result)
```

***3. Dictionary Comprehension from List***

```python
#Example
names = ["Ram", "Shyam", "Mohan"]
lengths = {name: len(name) for name in names}
print(lengths)
```

***4. Dictionary Comprehension from Dictionary***

```python
# Example

marks = {"Math": 80, "CS": 90, "Eng": 70}

updated = {k: v+5 for k, v in marks.items()}
print(updated)
```

***5 Filter Dictionary Using Comprehension***

```python
# Example
marks = {"Math": 80, "CS": 90, "Eng": 70}

passed = {k: v for k, v in marks.items() if v >= 75}
print(passed)
```

***6. Swap key and Values***

```python
# Example
data = {"a": 1, "b": 2, "c": 3}

swapped = {v: k for k, v in data.items()}
print(swapped)
```

***7. Dictionary Comprehension Using String***

```python
# Example
word = "python"
freq = {ch: word.count(ch) for ch in word}
print(freq)
```

***8. Dictionary Comprehension vs Loop***

| Feature     | For Loop    | Dict Comprehension |
| ----------- | ----------- | ------------------ |
| Code length | Long        | Short              |
| Readability | Medium      | High               |
| Speed       | Slower      | Faster             |
| Style       | Traditional | Pythonic           |


#### 9. Practice Set

```python
# Qns 1 : Square of numbers
d = {x: x*x for x in range(1, 6)}
print(d)


# Qns 2 : Even numbers and their squares
evens = {x: x*x for x in range(1, 11) if x % 2 == 0}
print(evens)

# Qns 3 : Even or Odd
result = {x: "Even" if x % 2 == 0 else "Odd" for x in range(1, 6)}
print(result)


# Qns 4 : Length of names
names = ["Ram", "Shyam", "Mohan"]
lengths = {name: len(name) for name in names}
print(lengths)


# Qns 5 : Update marks by adding grace marks
marks = {"Math": 80, "CS": 90, "Eng": 70}   
updated = {k: v+5 for k, v in marks.items()}
print(updated)

# Qns 6 : Character frequency in a word
word = "programming"
freq = {ch: word.count(ch) for ch in word}
print(freq)

# Qns 7 : Swapping keys and values in a dictionary
data = {"a": 1, "b": 2, "c": 3}
swapped = {v: k for k, v in data.items()}
print(swapped)


# Qns 8 : Filter dictionary based on value
marks = {"Math": 80, "CS": 90, "Eng": 70}   
passed = {k: v for k, v in marks.items() if v >= 75}
print(passed)

# Qns 9 : Create a dictionary from two lists
keys = ['a', 'b', 'c']
values = [1, 2, 3]
combined = {k: v for k, v in zip(keys, values)}
print(combined)

# Qns 10 : Convert List to dictionary
nums = [1, 2, 3, 4]
d = {x: x**3 for x in nums}
print(d)


# Qns 11 : Dictionary of multiples
multiples_of_3 = {x: x*3 for x in range(1, 11)}
print(multiples_of_3)

```

***10 Imp. Points***

* Dictionary comprehension creates dictionary in one line

* Syntax uses {key: value for ...}

* Supports conditions and if–else

* Faster and more readable than loops

* Used for transformation and filtering

* Values must be immutable when used as keys
