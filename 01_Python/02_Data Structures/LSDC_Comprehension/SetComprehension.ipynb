### Set Comprehension in Python

Set comprehension is a compact way to create a set using a single line of code, similar to list comprehension.

**key Points**

* Uses { }

* Stores unique elements only

* Automatically removes duplicates

* Faster and cleaner than loops


***Basic Syntax***

    new_set = {expression for item in iterable}


```python
# Example
s = {x for x in range(1, 6)}
print(s)
```

***Normal Set Creation Vs Comprehension***

```python
# Using for Loop
squares = set()
for i in range(1, 6):
    squares.add(i * i)

print(squares)
```

```python
# Using Set Comprehension
squares = {i * i for i in range(1, 6)}
print(squares)
```

***1. Set Comprehension With Comprehension***

***syntax***
   
   {expression for item in iterable if condition}



```python
# Example
evens = {x for x in range(1, 11) if x % 2 == 0}
print(evens)
```

***2. Set Comprehension with if-else***

In set comprehension, if–else must be inside the expression.


***Syntax***

   {expression_if_true if condition else expression_if_false for item in iterable}


```python
# Example

result = {"Even" if x % 2 == 0 else "Odd" for x in range(1, 6)}
print(result)
```

***3. Set Comprehension with List***

```python
# Example
nums = [1, 2, 2, 3, 4, 4, 5]

unique_nums = {x for x in nums}
print(unique_nums)
```

***4. Set Comprehension with String***

```python
# Example
word = "programming"
letters = {ch for ch in word}
print(letters)

#Unique characters only
```

***5. Set Comprehension with Function***

```python
# Example
def square(n):
    return n * n

result = {square(x) for x in range(1, 6)}
print(result)
```

***6. Set Comprehension with Nested Loop***

***Syntax***

    {expression for item1 in iterable1 for item2 in iterable2}


```python
# Example
pairs = {(x, y) for x in [1, 2] for y in [3, 4]}
print(pairs)
```

***7. Set Comprehension Vs List Comprehension***

| Feature          | List Comprehension | Set Comprehension |
| ---------------- | ------------------ | ----------------- |
| Brackets         | `[]`               | `{}`              |
| Duplicate values | Allowed            | Removed           |
| Order            | Preserved          | Not guaranteed    |
| Result type      | List               | Set               |


***8. Practice Sets***

```python
# Qns 1 : Square of Numbers
squares = {x**2 for x in range(1, 11)}
print(squares)
```

```python
# Qns 2 : Vowels in a String
string = "comprehensions in python"
vowels = {ch for ch in string if ch.lower() in 'aeiou'}
print(vowels)
```

```python
# Qns 3 : Removes Duplicates 
nums = [1, 2, 2, 3, 3, 4]
unique = {x for x in nums}
print(unique)
```

```python
# Qns 4 : First Letters of Words in set
words = {"apple", "banana", "cherry", "date"}
first_letters = {word[0] for word in words}
print(first_letters)
```

```python
# Qns  5: Lengths of Words
words = {"hello", "world", "python", "set"}
lengths = {len(word) for word in words}
print(lengths)
```
