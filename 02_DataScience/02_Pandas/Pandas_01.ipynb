**1. Series :**

A Pandas Series is like a column in a table.

It is a one-dimensional array holding data of any type.

* One-dimensional labeled array

* Like a single column in Excel

```python
# Example of series
s = pd.Series([1, 2, 3, 4, 5])
print(s)
```

***Lables***

If nothing else is specified, the values are labeled with their index number. First value has index 0, second value has index 1 etc.

This label can be used to access a specified value.

```python
print(s[0])  # Accessing first element of the series
```

***Create Lables***

With the index arguments, you can name your own lables.

```python
# Create your own lables
s = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])
print(s)    


a = [1, 7, 2]

myvar = pd.Series(a, index = ["x", "y", "z"])

print(myvar)


# When you have created labels, you can access an item by referring to the label instead of the position of the item.
print(myvar["y"])
```

```python
# Create a Simple Pandas Series from a Dictionary
calories = {"day1": 420, "day2": 380, "day3": 390};
myvar = pd.Series(calories)
print(myvar)
```

```python
# Create a Series using only data from "day1" and "day2":
import pandas as pd

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories, index = ["day1", "day2"])

print(myvar)
```

**DataFrame :**

A Pandas DataFrame is a 2 dimensional data structure, like a 2 dimensional array, or a table with rows and columns.



```python
# Create a simple pandas dataframe
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df = pd.DataFrame(data)
print(df)
```

***Locate Row***

As you can see from the result above, the DataFrame is like a table with rows and columns.

Pandas use the loc attribute to return one or more specified row(s)

```python
#Return row 0:
#refer to the row index:
print(df.loc[0])
```

```python
#Return row 0 and 1:
#use a list of indexes:
print(df.loc[[0, 1]])
```

***Named Indexes***

With the index argument, you can name your own indexes.

```python
#Add a list of names to give each row a name:
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}
df = pd.DataFrame(data, index = ["day1", "day2", "day3"])
print(df)
```

***Locate Named Indexes***

Use the named index in the loc attribute to return the specified row(s).

```python
#Return "day2":
#refer to the named index:
print(df.loc["day2"])
```

**3. Read CSV File**

A simple way to store big data sets is to use CSV files (comma separated files).

CSV files contains plain text and is a well know format that can be read by everyone including Pandas.

In our examples we will be using a CSV file called 'data.csv'.

```python
# Load the CSV into a DataFrame:
df = pd.read_csv('DataSets/Data.csv')

# use to_string() to print the entire DataFrame.
print(df.to_string())
```

```python
# If you have a large DataFrame with many rows, Pandas will only return the first 5 rows, and the last 5 rows:

import pandas as pd

df = pd.read_csv('DataSets/Data.csv')

print(df) 

#In my system the number is 60, which means that if the DataFrame contains more than 60 rows, the print(df) statement will return only the headers and the first and last 5 rows.
```

**4. Exploring Data**

```python
# 1. head()  : Returns the first 5 rows of the DataFrame by default. You can specify the number of rows to return as an argument to the head() method.
result = df.head()
print(result)
```

```python
# 2. tail() : Returns the last 5 rows of the DataFrame by default. You can specify the number of rows to return as an argument to the tail() method.
result = df.tail()
print(result)
```

```python
# 3. shape() : Returns a tuple representing the dimensionality of the DataFrame. The first element of the tuple is the number of rows, and the second element is the number of columns.
result = df.shape
print(result)
```

```python
# 4. info() : Provides a concise summary of the DataFrame, including the number of non-null entries, data types of each column, and memory usage.
result = df.info()
print(result)
```

```python
# 5. describe() : Generates descriptive statistics of the DataFrame, such as count, mean, standard deviation, minimum, and maximum values for each numeric column.
result = df.describe()
print(result)
```

***Selcting Data***


```python
# Select one column:
result = df["Pulse"]
print(result)
```

```python
# Select Multiple colunns:
result = df[["Duration", "Pulse"]]
print(result)
```

```python

# Data Selection (Label Based)
result = df.loc[0] # Returns the first row of the DataFrame, which is indexed as 0. You can also use other index values to select different rows.
print(result)

result = df.loc[[0, 1]] # Returns the first and second rows of the DataFrame, which are indexed as 0 and 1. You can also use other index values to select different rows.
print(result)

```

```python
# Row Selection (Index Based)
result=df.iloc[0]
print(result)
```
