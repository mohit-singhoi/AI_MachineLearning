### Looping Statements in Python :

A loop is a control structure that repeatedly executes a block of code as long as a specified condition is satisfied or until a sequence ends.

Loops help reduce code repetition and automate repetitive tasks.


##### There are mainly three types of Loop in Python

1. For Loop
2. while Loop
3. Nested Loop


#### 1. for Loop 

The for loop is used to iterate over a sequence such as a list, tuple, string, or range. It is mainly used when the number of iterations is known in advance.


***Syntax :***

for variable in sequence:

      statements


```python
# Example 1 – Print numbers from 1 to 5

for i in range(1, 6):
    print(i)
```

```python
# Example 2 – Print even numbers from 2 to 10
for i in range(2, 11, 2):
    print(i)
    
    
```

```python
# Example 3 – Print elements of a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

#### 2. while Loop

The while loop executes a block of code repeatedly as long as the given condition is True. It is used when the number of iterations is not known beforehand.


***Syntax:***

while condition:

       statements


```python
# Example 1 – Print numbers from 1 to 5
i= 1
while i <= 5:
    print(i)
    i += 1
```

```python
# Example 2 – Print even numbers from 2 to 10
i=1
while i <= 10:
    if i % 2 == 0:
        print(i)
    i += 1
```

```python
# Example 3 – Print elements of a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

#### 3. Nested Loop 

A nested loop is a loop inside another loop. The inner loop executes completely for each iteration of the outer loop.

*** Syntax: ***



```python
# Example 1 – Print matrix pattern
rows = 3
cols = 4
for i in range(rows):
    for j in range(cols):
        print('*', end=' ')
    print()  # Move to the next line after each row
```

```python
# Example 2 - Print number in matrix pattern
rows = 3
cols = 4
num = 1
for i in range(rows):
    for j in range(cols):
        print(num, end=' ')
        num += 1
    print()  # Move to the next line after each row
```

```python
# Example 3
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

### Loop Control Statements

Loop control statements are used to change the normal execution flow of a loop. They allow the programmer to terminate a loop, skip iterations, or create placeholder blocks without stopping program execution.

These statements give better control over iteration behavior and improve program efficiency.

***Python provides three loop control statements:***

1. break

2. continue

3. pass

#### break Statement :

The break statement immediately terminates the loop and transfers control to the statement following the loop, regardless of the loop condition.

Once break is executed, the loop stops completely.


***Syntax***

break


***Working of break Stmt***

i. The loop starts execution.

ii. A condition inside the loop is checked.

iii. When the condition becomes True, the break statement executes.

iv. The loop stops immediately.

v. Program control moves outside the loop.

```python
# Example 1 -Stop loop when number is found
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
# Example 2 -Stop loop when number is found
numbers = [1, 3, 5, 7, 9, 11]
target = 7
for num in numbers:
    if num == target:
        print(f"Found the target number: {target}")
        break
    print(f"Checked number: {num}")
```

#### continue Statements :

The continue statement skips the remaining code of the current iteration and moves to the next iteration of the loop.

Unlike break, it does NOT stop the loop.


***Syntax***

continue

***Working of continue***

i. Loop starts execution.

ii. If continue condition becomes True:

      Remaining statements are skipped.

      Loop jumps to next iteration.

iii. Loop continues normally.

```python
# Example skip number

for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
# Example 2 - skip even numbers
i=1
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# Example 3 – Print elements of a list of element to skiping 'banana'
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    if fruit == "banana":
        continue
    print(fruit)
```
