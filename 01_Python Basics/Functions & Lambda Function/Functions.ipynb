### FUNCTIONS IN PYTHON (WITH CODE)

#### Function :

A function is a named block of code that performs a specific task and can be reused whenever needed.
Functions help reduce repetition, improve readability, and make programs modular.



***Why Function is Used***

Avoid writing the same code multiple times

Improve code reusability

Make programs easy to understand and maintain

Simplify debugging and testing

Divide large programs into smaller parts
