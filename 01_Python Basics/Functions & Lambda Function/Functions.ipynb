### FUNCTIONS IN PYTHON (WITH CODE)

#### Function :

A function is a named block of code that performs a specific task and can be reused whenever needed.
Functions help reduce repetition, improve readability, and make programs modular.



***Why Function is Used***

* Avoid writing the same code multiple times

* Improve code reusability

* Make programs easy to understand and maintain

* Simplify debugging and testing

* Divide large programs into smaller parts



#### Defining a Function

In Python, a function is defined using the def keyword, followed by the function name and parentheses.



```python
# Syntax
def function_name(parameters):
    statements
    return value
```

```python
# Example 1
def add(a, b):
    return a + b
result = add(3, 5)
print(result)  # Output: 8

```

```python
# Example 2
def greet(name="Guest"):
    return f"Hello, {name}!"
print(greet())          # Output: Hello, Guest!
print(greet("Alice"))   # Output: Hello, Alice!
```

#### Function with Parameters

Functions can accept input values called parameters.

```python
# Example
# Here, a and b are parameters, and the function returns their sum.
def add(a, b):
    return a + b

result = add(10, 5)
print(result)
```

##### User-Defined Functions

These functions are created by the programmer to perform specific tasks.

***Syntax***

def function_name(parameters):

     statements

     return value


***Explanation***

* def → Keyword used to define a function

* function_name → Name of the function (user-defined)

* parameters → Input values passed to the function (optional)

* statements → Code that performs the task

* return → Sends result back to the caller (optional)

```python
# Example

def square(num):

      return num * num

print(square(4))
```

##### Function Arguments

***Positional Arguments***

values are passed in Order

```python
# Example
def subtract(a, b):
    return a - b

print(subtract(10, 5))
```

***Keyword Arguments***

Arguments are passed using parameter names.


```python
# Example
print(subtract(b=5, a=10)) # Output: 5
```

***Default Arguments***

Default values are used if no argument is provided.


```python
# Example
def welcome(name="User"):
    print("Welcome", name)

welcome()
welcome("Mohit")
```

***variable-Length-Arguments***

Used when number of argumentsis known


```python
# Example
def total(*numbers):
    sum = 0
    for n in numbers:
        sum += n
    return sum

print(total(1, 2, 3, 4))
print(total(10, 20))
```

***Return Statement***

The return statement sends a value back to the caller.

* A function can return multiple values.


```python
# Example
def calculate(a, b):
    return a+b, a-b, a*b

result = calculate(10, 5)
print(result)
sum_result, diff_result, prod_result = calculate(10, 5)
print(sum_result)
print(diff_result)
print(prod_result)
# Output:
# (15, 5, 50)
# 15
# 5
# 50
```

### Variable Scope in Functions

* Local Variable

* Global Variable

***1. Local Vriable***

Defined inside a function and used only inside the function.


```python
# Example
def demo():
    x = 10
    print(x)

demo()
# Output: 10
```

***2. Global Variable***

Defined outside a function, and it is used any where of the program.

It define the varible globaly.

```python
# Example
x = 20

def show():
    print(x)

show()
# Output: 20
```
